# Cross-cohort inference, Danenberg → Jackson: one set of predictions, three metrics

The reverse direction of `cross-cohort-metrics`. Models are trained on Danenberg (or on neither cohort) and scored on the
Jackson test patients (replicate-0 split of `train-test-grid`).

**Metrics:** (a) pooled pixel r, (b) per-image pixel r (mean over images), (c) cell-level r with each image's mean removed (R4).

**Models:**
1. EVA, zero-shot (pretrained on neither cohort). No image model trained on Danenberg exists yet.
2. LightGBM ("Table"), MLP, GCN, GraphSAGE, trained on the Danenberg train patients (replicate-0 split and seeds of `train-test-grid`).
   - Main version `[Danenberg-scaled inputs]`: Jackson inputs scaled with Danenberg statistics, so no cohort-level adaptation to
     Jackson (as for the image models).
   - Also the version with Jackson's own cohort scaling.
3. LightGBM trained on the Jackson train patients, directly on the target-space truth: the within-cohort reference.
4. Painted true cell means: the oracle for any per-cell map on (a) and (b).

**Truth:** the same target space as `cross-cohort-metrics`, `clip(arcsinh(x/5) / p99_Jackson-train, 0, 1)`, built from the Kaggle
Jackson TIFFs. Cells are the published masks relabelled by `jackson-cells-prep`.

**Per-cell models in pixel space:** each cell's prediction fills its mask pixels, background 0; pixel r follows exactly from per-image
sums. Danenberg-trained predictions are mapped back with Danenberg statistics, then to target space.

**Uncertainty:** 1000× bootstrap over Jackson test patients (same draws for every model, metric and marker).

## Cell tables (Danenberg cells to analyse; Jackson cells for the baselines)

```python
P = {}
for p in g("cells.parquet"):
    key = "danenberg" if "metabric_id" in pq.read_schema(p).names else "jackson"
    P[key] = dict(cells=p, contacts=near("cell_neighbours.parquet", p), report=near("prep_report.json", p))
P["danenberg"].update(clinical=near("clinical.parquet", P["danenberg"]["cells"]),
                      split=near("danenberg_patient_split.csv", P["danenberg"]["cells"]))
print(json.dumps(P, indent=1))

def load_table(key):
    cells = pd.read_parquet(P[key]["cells"])
    if key == "danenberg":
        cells = cells[(cells.is_tumour == 1) & (cells.is_hot_aggregate == 0)]
        pid = "metabric_id"
    else:
        cells = cells[cells.is_tumour == 1]
        pid = "patient_id"
    n_per = cells.groupby("image_name").size()
    cells = cells[cells.image_name.isin(n_per.index[n_per >= MIN_CELLS])]
    cells = cells.sort_values(["image_name", "object_number"]).reset_index(drop=True)
    img_code, img_names = pd.factorize(cells.image_name, sort=True)
    starts = np.r_[0, np.flatnonzero(np.diff(img_code)) + 1, len(cells)]
    loc = pd.Series(np.arange(len(cells)), index=pd.MultiIndex.from_arrays(
        [cells.image_number.values, cells.object_number.values]))
    con = pd.read_parquet(P[key]["contacts"])
    ca = loc.reindex(pd.MultiIndex.from_arrays([con.ImageNumber.values, con["from"].values])).values
    cb = loc.reindex(pd.MultiIndex.from_arrays([con.ImageNumber.values, con["to"].values])).values
    ok = ~(np.isnan(ca) | np.isnan(cb))
    ca, cb = ca[ok].astype(np.int64), cb[ok].astype(np.int64)
    ca, cb = ca[ca != cb], cb[ca != cb]
    if key == "danenberg":
        epi_frac = cells.groupby("cell_phenotype").is_epithelial.mean()
        bmap = {p: "epithelial" if epi_frac[p] > 0.5 else
                "stromal" if any(s in p.lower() for s in STROMAL_KEYS) else "immune" for p in epi_frac.index}
        broad = cells.cell_phenotype.map(bmap).astype(str).values
        epi = cells.is_epithelial.values == 1
    else:
        broad = cells.broad_class.astype(str).values
        epi = broad == "epithelial"
    return dict(key=key, cells=cells, img=img_code, img_names=np.asarray(img_names), starts=starts,
                patient=cells[pid].astype(str).values, xy=cells[["x", "y"]].values.astype(np.float64),
                ca=ca, cb=cb, broad=broad, epi=epi, phen=cells.cell_phenotype.astype(str).values,
                log_area=np.log(np.maximum(cells["area"].values, 1)).astype(np.float32))

DD, DJ = load_table("danenberg"), load_table("jackson")
N = len(DD["cells"])
split = pd.read_csv(P["danenberg"]["split"]).set_index("metabric_id").split
DD["split"] = pd.Series(DD["patient"]).map(split).fillna("none").values
clin = pd.read_parquet(P["danenberg"]["clinical"]).replace({"nan": np.nan})
clin = clin.assign(pid=clin.metabric_id.astype(str)).groupby("pid").first()
print(f"Danenberg: {N:,} cells, {len(DD['img_names'])} images, {len(set(DD['patient']))} patients; "
      f"split cells {pd.Series(DD['split']).value_counts().to_dict()}")
print(f"Jackson: {len(DJ['cells']):,} cells, {len(DJ['img_names'])} images")
pats, pcode = np.unique(DD["patient"], return_inverse=True)
W = np.random.default_rng(SEED).multinomial(len(pats), np.full(len(pats), 1 / len(pats)), size=N_BOOT).astype(np.float64)
tick("tables loaded")

def contact_matrix(D):
    n = len(D["cells"])
    C = sp.coo_matrix((np.ones(len(D["ca"]), np.float32), (D["ca"], D["cb"])), shape=(n, n)).tocsr()
    C = (C + C.T).tocsr()
    C.data[:] = 1.0
    return C

def ring_means(X, D, C):
    """Mean of X over non-touching cells within RING_R of the centroid, and their count."""
    n, p = X.shape
    Mr = np.full((n, p), np.nan, np.float32)
    nr = np.zeros(n, np.float32)
    for a, b in zip(D["starts"][:-1], D["starts"][1:]):
        m = b - a
        pr = cKDTree(D["xy"][a:b]).query_pairs(RING_R, output_type="ndarray")
        R = sp.coo_matrix((np.ones(len(pr), np.float32), (pr[:, 0], pr[:, 1])), shape=(m, m)).tocsr()
        R = (R + R.T).tocsr()
        Ci = C[a:b, a:b].tocsr()
        R = (R - R.multiply(Ci)).tocsr()
        R.eliminate_zeros()
        dr = np.asarray(R.sum(1)).ravel()
        Mr[a:b] = np.where(dr[:, None] > 0, (R @ X[a:b]) / np.maximum(dr, 1)[:, None], np.nan)
        nr[a:b] = dr
    return Mr, nr
```

## Features and graphs (both cohorts)

```python
try:
    import torch
except ImportError:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "torch"], check=True)
    import torch
import torch.nn as nn
import torch.nn.functional as F
dev = "cuda" if torch.cuda.is_available() else "cpu"
K_NN = 10
EPOCHS = 30 if SMOKE else 400
PATIENCE = 10 if SMOKE else 40
HID = 32 if SMOKE else 128
BUDGET_MIN = float(os.environ.get("BUDGET_MIN", 660))
REPS = [int(r) for r in os.environ.get("REPS", "0,1,2").split(",")]
SOURCES = ["Jackson", "Danenberg", "Both"]
MODELS = ["Table", "MLP", "GCN", "GraphSAGE"]
OTHER = {"Jackson": "Danenberg", "Danenberg": "Jackson"}

def harm(X):
    med = np.nanmedian(X, 0)
    q05, q95 = np.nanpercentile(X, [5, 95], 0)
    sc = np.where(q95 - q05 > 1e-3, q95 - q05, np.nanstd(X, 0) + 1e-6)
    return ((X - med) / sc).astype(np.float32)

dix = {m: i for i, m in enumerate(D_CHAN)}
IN_SET = NUC + MARKERS
XD_in = np.arcsinh(RAW[:, [dix[c] for c in IN_SET]] / COF).astype(np.float32)
XJ_in = np.arcsinh(DJ["cells"][IN_SET].values.astype(np.float64) / COF).astype(np.float32)
H = {"Jackson": np.c_[harm(XJ_in), harm(DJ["log_area"][:, None])],
     "Danenberg": np.c_[harm(XD_in), harm(DD["log_area"][:, None])]}
nJ, nD = len(H["Jackson"]), len(H["Danenberg"])
print(f"non-finite targets: Jackson {int((~np.isfinite(XJ_in)).sum())}, Danenberg {int((~np.isfinite(XD_in)).sum())}")

def knn_adj(D):
    n = len(D["cells"])
    rows, cols = [], []
    for a, b in zip(D["starts"][:-1], D["starts"][1:]):
        k = min(K_NN, b - a - 1)
        _, nb = cKDTree(D["xy"][a:b]).query(D["xy"][a:b], k=k + 1)
        rows.append(np.repeat(np.arange(a, b), k))
        cols.append(nb[:, 1:].ravel() + a)
    A = sp.coo_matrix((np.ones(sum(map(len, rows)), np.float32), (np.concatenate(rows), np.concatenate(cols))),
                      shape=(n, n)).tocsr()
    return ((A + A.T) > 0).astype(np.float32).tocsr()

def normalise(A):
    n = A.shape[0]
    deg = np.asarray(A.sum(1)).ravel()
    A_mean = (sp.diags(1 / np.maximum(deg, 1)) @ A).tocsr()                # GraphSAGE mean over neighbours
    Ai = A + sp.eye(n, dtype=np.float32, format="csr")
    dh = 1 / np.sqrt(np.asarray(Ai.sum(1)).ravel())
    A_gcn = (sp.diags(dh) @ Ai @ sp.diags(dh)).tocsr()                     # Kipf normalisation with self-loops
    return {"mean": A_mean, "gcn": A_gcn}, float(deg.mean())

ADJ = {}
ADJ["Jackson"], degJ = normalise(knn_adj(DJ))
ADJ["Danenberg"], degD = normalise(knn_adj(DD))
print(f"kNN graphs: mean degree Jackson {degJ:.1f}, Danenberg {degD:.1f}; device {dev}")

def to_csr(M):
    M = M.tocsr().astype(np.float32)
    M.sort_indices()
    return torch.sparse_csr_tensor(torch.from_numpy(M.indptr.astype(np.int32)), torch.from_numpy(M.indices.astype(np.int32)),
                                   torch.from_numpy(M.data), size=M.shape).to(dev)

class _SpMM(torch.autograd.Function):
    """A @ X with a precomputed CSR transpose for the backward pass."""
    @staticmethod
    def forward(ctx, X, A, AT):
        ctx.AT = AT
        return A @ X
    @staticmethod
    def backward(ctx, G):
        return ctx.AT @ G.contiguous(), None, None

def spmm(Ab, X):
    return _SpMM.apply(X, Ab[0], Ab[1])

def build_graph(src):
    """GPU tensors for a training/prediction graph: features, A (and A^T), row sums, and A @ features."""
    if src == "Both":
        mats = {k: sp.block_diag([ADJ["Jackson"][k], ADJ["Danenberg"][k]], format="csr") for k in ("mean", "gcn")}
        Xn = np.r_[H["Jackson"], H["Danenberg"]]
    else:
        mats, Xn = ADJ[src], H[src]
    Xn = np.nan_to_num(Xn)                                                 # NaN -> 0, the cohort median
    gr = {"x": torch.from_numpy(Xn).to(dev)}
    for k, M in mats.items():
        A = to_csr(M)
        gr[k] = (A, A if k == "gcn" else to_csr(M.T))
        gr["a1_" + k] = torch.from_numpy(np.asarray(M.sum(1)).ravel().astype(np.float32)).to(dev)
        gr["ax_" + k] = torch.from_numpy((M @ Xn).astype(np.float32)).to(dev)
    return gr

GR = {}
def ensure(keys):
    """Keep only the graphs in keys on the GPU (the pooled graph is large)."""
    for k in [k for k in GR if k not in keys]:
        del GR[k]
    if dev == "cuda":
        torch.cuda.empty_cache()
    for k in keys:
        if k not in GR:
            GR[k] = build_graph(k)

ensure({"Jackson", "Danenberg"})
tick("features and graphs")
```

## Models (identical maths to `model-comparison`; first-layer aggregation precomputed)

```python
from torch.utils.checkpoint import checkpoint
CKPT = {"on": False}       # gradient checkpointing (memory only; same maths), used for the pooled graph

def run(fn, *a):
    return checkpoint(fn, *a, use_reentrant=False) if (CKPT["on"] and torch.is_grad_enabled()) else fn(*a)

class MLP(nn.Module):
    def __init__(s, d, h=HID, p=0.1):
        super().__init__()
        s.net = nn.Sequential(nn.Linear(d, h), nn.ReLU(), nn.Dropout(p), nn.Linear(h, h), nn.ReLU(), nn.Dropout(p),
                              nn.Linear(h, 1))
    def forward(s, x, ax, A, a1):
        return s.net(x).squeeze(-1)

class GCN(nn.Module):
    def __init__(s, d, h=HID, p=0.1):
        super().__init__()
        s.l1, s.l2, s.out, s.dp = nn.Linear(d, h), nn.Linear(h, h), nn.Linear(h, 1), nn.Dropout(p)
    def forward(s, x, ax, A, a1):
        # layer 1: A (x W1 + b1) = (A x) W1 + (A 1) b1, as one product
        l1 = lambda ax, a1: s.dp(torch.relu(F.linear(torch.cat([ax, a1[:, None]], 1),
                                                     torch.cat([s.l1.weight, s.l1.bias[:, None]], 1))))
        l2 = lambda h: torch.relu(spmm(A, s.l2(h)))
        return s.out(run(l2, run(l1, ax, a1))).squeeze(-1)

class SAGE(nn.Module):
    def __init__(s, d, h=HID, p=0.1):
        super().__init__()
        s.s1, s.n1 = nn.Linear(d, h), nn.Linear(d, h, bias=False)
        s.s2, s.n2 = nn.Linear(h, h), nn.Linear(h, h, bias=False)
        s.out, s.dp = nn.Linear(h, 1), nn.Dropout(p)
    def forward(s, x, ax, A, a1):
        # layer 1: x S1 + A (x N1) = [x, A x] [S1; N1], as one product
        l1 = lambda x, ax: s.dp(torch.relu(F.linear(torch.cat([x, ax], 1), torch.cat([s.s1.weight, s.n1.weight], 1),
                                                    s.s1.bias)))
        l2 = lambda h: torch.relu(s.s2(h) + spmm(A, s.n2(h)))
        return s.out(run(l2, run(l1, x, ax))).squeeze(-1)

NETS = {"MLP": (MLP, "mean"), "GCN": (GCN, "gcn"), "GraphSAGE": (SAGE, "mean")}

def net_inputs(gr, akey, cols):
    return gr["x"][:, cols], gr["ax_" + akey][:, cols], gr[akey], gr["a1_" + akey]

def train_net(cls, inp, y, tr, va, seed):
    torch.manual_seed(seed)
    m = cls(inp[0].shape[1]).to(dev)
    opt = torch.optim.Adam(m.parameters(), lr=3e-3, weight_decay=1e-5)
    tr, va = torch.from_numpy(tr).to(dev), torch.from_numpy(va).to(dev)
    best, state, bad, ep = np.inf, None, 0, 0
    for ep in range(EPOCHS):
        m.train()
        opt.zero_grad()
        loss = ((m(*inp)[tr] - y[tr]) ** 2).mean()
        loss.backward()
        opt.step()
        if ep % 5 == 4:
            m.eval()
            with torch.no_grad():
                v = ((m(*inp)[va] - y[va]) ** 2).mean().item()
            if v < best - 1e-5:
                best, bad = v, 0
                state = {k: t.detach().clone() for k, t in m.state_dict().items()}
            else:
                bad += 5
                if bad >= PATIENCE:
                    break
    if state is not None:
        m.load_state_dict(state)
    m.eval()
    return m, ep + 1, best
```

## Scoring helpers

```python
def stats_fn(img, pcode, n_pat, W, center):
    def f(a, b):
        ok = np.isfinite(a) & np.isfinite(b)
        a, b, im, pc = a[ok].astype(np.float64), b[ok].astype(np.float64), img[ok], pcode[ok]
        if center:
            k = int(img.max()) + 1
            cnt = np.maximum(np.bincount(im, minlength=k), 1)
            a = a - (np.bincount(im, a, k) / cnt)[im]
            b = b - (np.bincount(im, b, k) / cnt)[im]
        else:
            a, b = a - a.mean(), b - b.mean()
        sxy, sxx, syy = (np.bincount(pc, a * b, n_pat), np.bincount(pc, a * a, n_pat), np.bincount(pc, b * b, n_pat))
        return float(sxy.sum() / np.sqrt(sxx.sum() * syy.sum())), (W @ sxy) / np.sqrt((W @ sxx) * (W @ syy))
    return f

def ci(v):
    v = np.asarray(v)[np.isfinite(v)]
    return float(np.percentile(v, 2.5)), float(np.percentile(v, 97.5))

def scorers(D, te, seed):
    tp, tcode = np.unique(D["patient"][te], return_inverse=True)
    Wt = np.random.default_rng(seed).multinomial(len(tp), np.full(len(tp), 1 / len(tp)), size=N_BOOT).astype(np.float64)
    return {"R4": stats_fn(D["img"][te], tcode, len(tp), Wt, True), "r": stats_fn(D["img"][te], tcode, len(tp), Wt, False)}, len(tp)

# Danenberg: frozen test patients (fixed across replicates)
D_TE = np.flatnonzero(DD["split"] == "test")
D_TRVA_PAT = np.unique(DD["patient"][DD["split"] != "test"])
SC_D, n_tpD = scorers(DD, D_TE, SEED + 13)
TRUTH_D = {m: {"cell mean": XD_in[D_TE, IN_SET.index(m)], "target map": TRUTH[m][D_TE]} for m in MARKERS}
print(f"Danenberg test: {len(D_TE):,} cells, {n_tpD} patients")
```

## Splits (replicate 0), Jackson images and masks

```python
REP = 0
jp, jcode = np.unique(DJ["patient"], return_inverse=True)
u = np.random.default_rng(SEED + 11 + REP).random(len(jp))
J_SPLIT = np.where(u < 0.7, "train", np.where(u < 0.8, "val", "test"))[jcode]
uv = np.random.default_rng(SEED + 21 + REP).random(len(D_TRVA_PAT))
dval = set(D_TRVA_PAT[uv < 0.125])
D_SPLIT = np.where(DD["split"] == "test", "test",
                   np.where(pd.Series(DD["patient"]).isin(dval).values, "val", "train"))
D_TR, D_VA = np.flatnonzero(D_SPLIT == "train"), np.flatnonzero(D_SPLIT == "val")

tifs = g("*.tiff")
J_IMG = {os.path.splitext(os.path.basename(p))[0]: p for p in tifs
         if not os.path.basename(p).startswith("MB_") and "/masks/" not in p.replace("\\", "/") and "jackson" in p.lower()}
J_MASK = {os.path.splitext(os.path.basename(p))[0]: p for p in tifs
          if not os.path.basename(p).startswith("MB_") and "/masks/" in p.replace("\\", "/")}
J_NAMES = list(DJ["img_names"])
JOBJ = DJ["cells"].object_number.values.astype(np.int64)
JST = DJ["starts"]
HAVE = np.array([s in J_IMG and s in J_MASK for s in J_NAMES])
print(f"Jackson images with a tiff and a mask: {HAVE.sum()} / {len(J_NAMES)} (tiffs {len(J_IMG)}, masks {len(J_MASK)})")

nJ = len(DJ["cells"])
J_TR = np.flatnonzero(J_SPLIT == "train")
J_TE = np.flatnonzero((J_SPLIT == "test") & HAVE[DJ["img"]])
print(f"Danenberg train {len(D_TR):,} / val {len(D_VA):,} cells; Jackson test {len(J_TE):,} cells in "
      f"{len(np.unique(DJ['img'][J_TE]))} images, {len(set(DJ['patient'][J_TE]))} patients")

jix = {m: i for i, m in enumerate(J_CHAN)}
CACHE = os.path.dirname(g("channel_preproc_residual.pkl")[0])
_, TNS = pickle.load(open(f"{CACHE}/channel_preproc_residual.pkl", "rb"))

def to_target(m, arc):
    p01, p99 = TNS[jix[m]]
    return np.clip((arc - p01) / (p99 - p01 + 1e-8), 0, 1).astype(np.float32)

def read_hwc(path, n_chan):
    im = tifffile.imread(path)
    if im.ndim == 3 and im.shape[0] == n_chan and im.shape[-1] != n_chan:
        im = np.moveaxis(im, 0, -1)
    assert im.shape[-1] == n_chan, (path, im.shape)
    return im.astype(np.float32)

def truth_map_j(im_hwc, m):
    """Target space of the image models: clip(arcsinh(x/5) / p99_Jackson, 0, 1) (p01 = 0)."""
    p01, p99 = TNS[jix[m]]
    return np.clip((np.arcsinh(im_hwc[..., jix[m]] / COF) - p01) / (p99 - p01 + 1e-8), 0., 1.)

def loc_from(mask_path, obj):
    """Mask pixel -> cell index within the image (-1 = background or unmatched)."""
    mask = tifffile.imread(mask_path).astype(np.int64)
    lut = np.full(max(int(mask.max()), int(obj.max())) + 1, -1, np.int64)
    lut[obj] = np.arange(len(obj))
    return lut[mask], mask.shape

def j_loc(i):
    return loc_from(J_MASK[J_NAMES[i]], JOBJ[JST[i]:JST[i + 1]])
```

## Jackson truth: target-space maps, cell means and pixel sums from the TIFFs and relabelled masks

```python
def j_truth(name, obj, img_path, mask_path):
    """Truth cell means (target space), cell pixel counts and per-image truth sums for one Jackson image."""
    n = len(obj)
    if img_path is None:
        return np.zeros(n, np.int32), {m: np.full(n, np.nan, np.float32) for m in MARKERS}, []
    loc, shape = loc_from(mask_path, obj)
    im = read_hwc(img_path, len(J_CHAN))
    assert im.shape[:2] == shape, (name, im.shape, shape)
    loc = loc.ravel()
    v = loc >= 0
    cnt = np.bincount(loc[v], minlength=n)
    cm, sums = {}, []
    for m in MARKERS:
        G = truth_map_j(im, m).ravel().astype(np.float64)
        with np.errstate(invalid="ignore", divide="ignore"):
            cm[m] = np.where(cnt > 0, np.bincount(loc[v], G[v], n) / np.maximum(cnt, 1), np.nan).astype(np.float32)
        sums.append(dict(marker=m, image=name, n=G.size, Sg=G.sum(), Sgg=G @ G))
    return cnt.astype(np.int32), cm, sums

_jt = Parallel(n_jobs=4)(delayed(j_truth)(s, JOBJ[JST[i]:JST[i + 1]], J_IMG[s] if HAVE[i] else None,
                                          J_MASK[s] if HAVE[i] else None) for i, s in enumerate(J_NAMES))
NCELLJ = np.concatenate([c for c, _, _ in _jt])
TRUTHJ = {m: np.concatenate([t[m] for _, t, _ in _jt]) for m in MARKERS}
TRUTH_SUMS = pd.DataFrame([r for _, _, s in _jt for r in s]).set_index(["marker", "image"])
del _jt
assert len(NCELLJ) == nJ
print(f"cells with mask pixels: {np.mean(NCELLJ > 0):.4f}")
# the truth cell means must agree with the published Jackson cell table (checks mask relabelling and channel order)
chk = []
for m in MARKERS:
    t, x = TRUTHJ[m], XJ_in[:, IN_SET.index(m)]
    ok = np.isfinite(t) & np.isfinite(x)
    chk.append(dict(marker=m, r_truth_vs_table=float(np.corrcoef(t[ok], x[ok])[0, 1]), n=int(ok.sum())))
chk = pd.DataFrame(chk)
print(chk.round(3).to_string(index=False))
chk.to_csv(f"{OUT}/truth_vs_table_check.csv", index=False)
assert chk.r_truth_vs_table.median() > 0.6, "truth cell means disagree with the Jackson table"
tick("Jackson truth maps, cell means and pixel sums")
```

## Train on Danenberg (per-cell models) and on Jackson (reference); predict the Jackson test cells

```python
# Danenberg harmonisation statistics (the numbers harm() used for H["Danenberg"])
D_MED = np.nanmedian(XD_in, 0)
_q05, _q95 = np.nanpercentile(XD_in, [5, 95], 0)
D_SC = np.where(_q95 - _q05 > 1e-3, _q95 - _q05, np.nanstd(XD_in, 0) + 1e-6)
_chk = np.random.default_rng(0).choice(len(XD_in), 20_000, replace=False)
assert np.allclose((XD_in[_chk] - D_MED) / D_SC, H["Danenberg"][_chk, :XD_in.shape[1]], atol=1e-4, equal_nan=True)
_la = DD["log_area"].astype(np.float64)
LA_MED = np.nanmedian(_la)
_l05, _l95 = np.nanpercentile(_la, [5, 95])
LA_SC = _l95 - _l05 if _l95 - _l05 > 1e-3 else np.nanstd(_la) + 1e-6
assert np.allclose((_la[_chk] - LA_MED) / LA_SC, H["Danenberg"][_chk, -1], atol=1e-4, equal_nan=True)

# main version: Jackson inputs scaled with the Danenberg statistics (no cohort-level adaptation to Jackson)
JS = "Jackson (Danenberg-scaled)"
H[JS] = np.c_[(XJ_in - D_MED) / D_SC, (DJ["log_area"] - LA_MED) / LA_SC].astype(np.float32)
ADJ[JS] = ADJ["Jackson"]
ensure({"Jackson", "Danenberg", JS})
CTRL = " [Danenberg-scaled inputs]"
PERCELL = ["Table", "MLP", "GCN", "GraphSAGE"]

HARMP, TSP, LOG = {}, {}, []
Hs = H["Danenberg"]
for m in MARKERS:
    k = IN_SET.index(m)
    cols = [i for i in range(Hs.shape[1]) if i != k]
    y_all = Hs[:, k]
    trm, vam = D_TR[np.isfinite(y_all[D_TR])], D_VA[np.isfinite(y_all[D_VA])]
    mu, sd = y_all[trm].mean(), y_all[trm].std() + 1e-6
    fit = np.sort(np.random.default_rng(SEED + REP).choice(trm, min(N_FIT_FINAL, len(trm)), replace=False))
    t0 = time.time()
    mdl = lgb.LGBMRegressor(random_state=SEED + REP, **LGB).fit(Hs[fit][:, cols], y_all[fit])
    HARMP[("Table", m)] = mdl.predict(H["Jackson"][J_TE][:, cols]).astype(np.float32)
    HARMP[("Table" + CTRL, m)] = mdl.predict(H[JS][J_TE][:, cols]).astype(np.float32)
    LOG.append(dict(model="Table", marker=m, epochs=np.nan, seconds=time.time() - t0, n_train=len(fit)))
    yt = torch.from_numpy(np.nan_to_num((y_all - mu) / sd)).to(dev)
    for name in ("MLP", "GCN", "GraphSAGE"):
        cls, akey = NETS[name]
        t0 = time.time()
        net, n_ep, best = train_net(cls, net_inputs(GR["Danenberg"], akey, cols), yt, trm, vam, SEED + REP)
        with torch.no_grad():
            pj = net(*net_inputs(GR["Jackson"], akey, cols)).cpu().numpy() * sd + mu
            pjs = net(*net_inputs(GR[JS], akey, cols)).cpu().numpy() * sd + mu
        HARMP[(name, m)] = pj[J_TE].astype(np.float32)
        HARMP[(name + CTRL, m)] = pjs[J_TE].astype(np.float32)
        LOG.append(dict(model=name, marker=m, epochs=n_ep, seconds=time.time() - t0, n_train=len(trm)))
        del net
        if dev == "cuda":
            torch.cuda.empty_cache()
    for name in PERCELL:
        for nm in (name, name + CTRL):
            TSP[(nm, m)] = to_target(m, HARMP[(nm, m)] * D_SC[k] + D_MED[k])
    # LightGBM, Jackson-trained, directly on the target-space truth cell means of the Jackson train patients
    yJ = TRUTHJ[m]
    trJ = J_TR[np.isfinite(yJ[J_TR])]
    fitJ = np.sort(np.random.default_rng(SEED + REP).choice(trJ, min(N_FIT_FINAL, len(trJ)), replace=False))
    t0 = time.time()
    mdlJ = lgb.LGBMRegressor(random_state=SEED + REP, **LGB).fit(H["Jackson"][fitJ][:, cols], yJ[fitJ])
    TSP[("Table (Jackson-trained)", m)] = np.clip(mdlJ.predict(H["Jackson"][J_TE][:, cols]), 0, 1).astype(np.float32)
    LOG.append(dict(model="Table (Jackson-trained)", marker=m, epochs=np.nan, seconds=time.time() - t0, n_train=len(fitJ)))
    TSP[("Painted true cell means", m)] = TRUTHJ[m][J_TE]
    tick(f"trained {m}")
pd.DataFrame(LOG).to_csv(f"{OUT}/training_log.csv", index=False)
MODEL_NAMES = PERCELL + [p + CTRL for p in PERCELL] + ["Table (Jackson-trained)", "Painted true cell means"]
```

## EVA, zero-shot, on the Jackson test images

```python
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "omegaconf", "einops"], check=True)
EVA_DIR = "/tmp/Eva"
if not os.path.isdir(EVA_DIR):
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/YAndrewL/Eva.git", EVA_DIR], check=True)
EVA_COMMIT = subprocess.run(["git", "-C", EVA_DIR, "rev-parse", "HEAD"], capture_output=True, text=True).stdout.strip()
os.makedirs(f"{EVA_DIR}/marker_embeddings", exist_ok=True)
gp = g("GenePT_embedding.pkl")[0]
if not os.path.exists(f"{EVA_DIR}/marker_embeddings/GenePT_embedding.pkl"):
    os.symlink(gp, f"{EVA_DIR}/marker_embeddings/GenePT_embedding.pkl")
os.chdir(EVA_DIR)                                # mae.py opens marker_embeddings/ relative to the working directory
sys.path.insert(0, EVA_DIR)
from omegaconf import OmegaConf
from Eva.eva import EvaMAE
from utils.constant import marker_to_gene
conf = OmegaConf.load(f"{EVA_DIR}/config.yaml")
eva = EvaMAE.from_checkpoint(g("Eva_model.ckpt")[0], conf, device=dev)
eva.eval()
os.chdir("/kaggle/working")
print("EVA commit", EVA_COMMIT, "| parameters", sum(p.numel() for p in eva.parameters()) / 1e6, "M")

EVA_NAME = {"histone_h3": "HistoneH3", "c_erb_b_2_her2": "HER2", "cd20": "CD20", "cd3": "CD3e", "cd45": "CD45",
            "cd68": "CD68", "cytokeratin_5": "Keratin5", "cytokeratin_8_18": "Keratin8/18",
            "estrogen_receptor_alpha": "ERa", "ki_67": "Ki67", "pan_cytokeratin": "PanCK", "sma": "aSMA"}
assert all(v in marker_to_gene for v in EVA_NAME.values()) and "DAPI" in marker_to_gene
CH_EVA = ["DAPI"] + [EVA_NAME[c] for c in ["histone_h3"] + MARKERS]       # 13 channels
TILE, TOK = 224, 8
NTOK = (TILE // TOK) ** 2

def eva_channels(im_hwc, idx, mode):
    """[H, W, 13] in [0, 1]: DAPI (mean of DNA1, DNA2), HistoneH3, the 11 markers; per-image, per-channel scaling."""
    raw = [0.5 * (im_hwc[..., idx["dna1"]] + im_hwc[..., idx["dna2"]])] + [im_hwc[..., idx[c]] for c in ["histone_h3"] + MARKERS]
    out = []
    for c in raw:
        c = np.arcsinh(c / COF) if mode == "arcsinh" else c
        lo, hi = np.percentile(c, [0.5, 99.5])
        out.append(np.clip((c - lo) / (hi - lo + 1e-8), 0, 1))
    return np.stack(out, -1).astype(np.float32)

@torch.no_grad()
def eva_predict_arr(x):
    """x: [H, W, 13] scaled. Returns {marker: map}, each marker predicted with its own channel hidden."""
    H, W, C = x.shape
    Hp, Wp = -(-H // TILE) * TILE, -(-W // TILE) * TILE
    xp = np.zeros((Hp, Wp, C), np.float32)
    xp[:H, :W] = x
    tiles = [(y0, x0) for y0 in range(0, Hp, TILE) for x0 in range(0, Wp, TILE)]
    preds = {}
    for k, m in enumerate(MARKERS):
        ci = 2 + k                                  # channel of the target marker
        mask = torch.zeros(C, NTOK, device=dev)
        mask[ci] = 1.0
        P = np.zeros((Hp, Wp), np.float32)
        for b0 in range(0, len(tiles), 16):
            tb = tiles[b0:b0 + 16]
            xb = np.stack([xp[y0:y0 + TILE, x0:x0 + TILE] for y0, x0 in tb])
            xb[..., ci] = 0.0
            # EVA permutes [B, H, W, C] -> [B, C, H, W] and then calls .view, so the memory must be [B, C, H, W]
            xb = torch.from_numpy(np.ascontiguousarray(xb.transpose(0, 3, 1, 2))).to(dev).permute(0, 2, 3, 1)
            with autocast(device_type="cuda", dtype=torch.float16, enabled=dev == "cuda"):
                rec, _, _ = eva(xb, marker_in=[CH_EVA] * len(tb), marker_out=[[CH_EVA[ci]]] * len(tb),
                                infer_mask=mask, channel_mask=None)
            rec = rec.float()[..., 0].cpu().numpy()
            for (y0, x0), r in zip(tb, rec):
                P[y0:y0 + TILE, x0:x0 + TILE] = r
        preds[m] = P[:H, :W]
    return preds

from torch.amp import autocast
```

```python
MODE = "arcsinh"            # chosen on Jackson val images in eva-zeroshot (0.163 vs 0.159 for linear)
IMG_TEJ = DJ["img"][J_TE]
TEST_IMGS_J = np.unique(IMG_TEJ)
EVA_ROWS = []
CELL_EVA = {m: np.full(len(J_TE), np.nan, np.float32) for m in MARKERS}

@guard("EVA")
def _():
    for j, i in enumerate(TEST_IMGS_J):
        s = J_NAMES[i]
        im = read_hwc(J_IMG[s], len(J_CHAN))
        preds = eva_predict_arr(eva_channels(im, jix, MODE))
        loc, shape = j_loc(i)
        loc = loc.ravel()
        v = loc >= 0
        pos = np.searchsorted(J_TE, JST[i] + loc[v])                 # test-array index of each cell pixel
        assert (J_TE[pos] == JST[i] + loc[v]).all()
        n_te = len(J_TE)
        cnt = np.bincount(pos, minlength=n_te)
        got = cnt > 0
        for m in MARKERS:
            G = truth_map_j(im, m).astype(np.float64).ravel()
            P = preds[m].astype(np.float64).ravel()
            assert P.size == G.size
            ts = TRUTH_SUMS.loc[(m, s)]
            assert abs(G.sum() - ts.Sg) <= 1e-6 * max(1.0, ts.Sg) and G.size == ts.n
            EVA_ROWS.append(dict(model="EVA", marker=m, image=s, n=G.size, Sg=G.sum(), Sgg=G @ G, Sp=P.sum(), Spp=P @ P,
                                 Sgp=G @ P))
            sm = np.bincount(pos, P[v], n_te)
            CELL_EVA[m][got] = (sm[got] / cnt[got]).astype(np.float32)
        if j % 20 == 0:
            tick(f"EVA: scored {j + 1}/{len(TEST_IMGS_J)} Jackson test images")
    for m in MARKERS:
        TSP[("EVA", m)] = CELL_EVA[m]
    MODEL_NAMES.append("EVA")
```

## Per-image pixel sums (painted per-cell maps) and the direct pixel check

```python
K_IMG = len(J_NAMES)
name_te = np.asarray(J_NAMES)[TEST_IMGS_J]
nc = NCELLJ[J_TE].astype(np.float64)
rows = []
for m in MARKERS:
    gs = TRUTH_SUMS.loc[m].reindex(name_te)
    gc_ = np.nan_to_num(TRUTHJ[m][J_TE].astype(np.float64))
    for M in MODEL_NAMES:
        if M == "EVA":
            continue
        p = np.nan_to_num(TSP[(M, m)].astype(np.float64))
        Sp = np.bincount(IMG_TEJ, nc * p, K_IMG)[TEST_IMGS_J]
        Spp = np.bincount(IMG_TEJ, nc * p * p, K_IMG)[TEST_IMGS_J]
        Sgp = np.bincount(IMG_TEJ, nc * gc_ * p, K_IMG)[TEST_IMGS_J]
        rows.append(pd.DataFrame(dict(model=M, marker=m, image=name_te, n=gs.n.values, Sg=gs.Sg.values,
                                      Sgg=gs.Sgg.values, Sp=Sp, Spp=Spp, Sgp=Sgp)))
SUMS_TE = pd.concat(rows + ([pd.DataFrame(EVA_ROWS)] if EVA_ROWS else []), ignore_index=True)
SUMS_TE.to_csv(f"{OUT}/pixel_sums_test.csv", index=False)

def r_from(n, Sg, Sp, Sgg, Spp, Sgp):
    with np.errstate(invalid="ignore", divide="ignore"):
        return (Sgp - Sg * Sp / n) / np.sqrt((Sgg - Sg ** 2 / n) * (Spp - Sp ** 2 / n))

# check: paint one prediction on one test image and compute pixel r directly
@guard("direct pixel check")
def _():
    i = TEST_IMGS_J[np.argmax(np.bincount(IMG_TEJ)[TEST_IMGS_J])]
    s = J_NAMES[i]
    im = read_hwc(J_IMG[s], len(J_CHAN))
    loc, _ = j_loc(i)
    pos = np.where(loc >= 0, np.searchsorted(J_TE, JST[i] + np.maximum(loc, 0)), -1)
    out = []
    for m in ("pan_cytokeratin", "cd3"):
        G = truth_map_j(im, m).astype(np.float64)
        for M in ("Table" + CTRL, "Table (Jackson-trained)"):
            pc = np.nan_to_num(TSP[(M, m)])
            Pm = np.where(pos >= 0, pc[np.maximum(pos, 0)], 0.0)
            r_dir = float(np.corrcoef(G.ravel(), Pm.ravel())[0, 1])
            row = SUMS_TE[(SUMS_TE.model == M) & (SUMS_TE.marker == m) & (SUMS_TE.image == s)].iloc[0]
            out.append(dict(image=s, marker=m, model=M, r_direct=r_dir,
                            r_sums=float(r_from(row.n, row.Sg, row.Sp, row.Sgg, row.Spp, row.Sgp))))
    out = pd.DataFrame(out)
    out.to_csv(f"{OUT}/direct_pixel_check.csv", index=False)
    print(out.round(5).to_string(index=False))
    assert (out.r_direct - out.r_sums).abs().max() < 1e-3

np.savez_compressed(f"{OUT}/percell_predictions_test.npz",
                    image=np.asarray(J_NAMES)[IMG_TEJ], object_number=JOBJ[J_TE],
                    patient=np.asarray(DJ["patient"][J_TE]).astype(str), n_pixels=NCELLJ[J_TE],
                    **{f"truth__{m}": TRUTHJ[m][J_TE] for m in MARKERS},
                    **{f"pred__{M}__{m}": TSP[(M, m)] for M in MODEL_NAMES for m in MARKERS})
tick("per-image pixel sums and per-cell predictions saved")
```

```python
PATCH = 32                                        # 32 µm patches (1 µm/px), non-overlapping, partial edge tiles dropped
PATCH_CHECK = []

def patch_means(A):
    h, w = (A.shape[0] // PATCH) * PATCH, (A.shape[1] // PATCH) * PATCH
    return A[:h, :w].reshape(h // PATCH, PATCH, w // PATCH, PATCH).mean((1, 3)).ravel()

def patch_row(MODEL, m, s, G2, P2, T2=None):
    """Per-image sums over 32 µm patch means (truth G2, raw prediction P2, target-scale prediction T2)."""
    g, p = patch_means(G2), patch_means(P2)
    row = dict(model=MODEL, marker=m, image=s, n=g.size, Sg=g.sum(), Sgg=g @ g, Sp=p.sum(), Spp=p @ p, Sgp=g @ p)
    if T2 is not None:
        t = patch_means(T2)
        row.update(Sp_t=t.sum(), Spp_t=t @ t, Sgp_t=g @ t)
    if len(PATCH_CHECK) < 4 and g.size > 2 and g.std() > 0 and p.std() > 0:   # sums must give the direct patch r
        PATCH_CHECK.append(abs(float(np.corrcoef(g, p)[0, 1]) - float(r_from(*(row[c] for c in ["n", "Sg", "Sp", "Sgg", "Spp", "Sgp"])))))
    return row

# Per-cell models and the oracle, painted into the masks (background 0) in target space, as for the pixel sums.
# Their predictions are already in target space, so the target-scale sums (_t) equal the raw ones.
# EVA's patch sums come from image-models-jackson (its figure source).
@guard("patch sums")
def _():
    rows = []
    for j, i in enumerate(TEST_IMGS_J):
        s = J_NAMES[i]
        im = read_hwc(J_IMG[s], len(J_CHAN))
        loc, _ = j_loc(i)
        pos = np.where(loc >= 0, np.searchsorted(J_TE, JST[i] + np.maximum(loc, 0)), -1)
        for m in MARKERS:
            G = truth_map_j(im, m).astype(np.float64)
            if j < 3:
                ts = TRUTH_SUMS.loc[(m, s)]
                assert abs(G.sum() - ts.Sg) <= 1e-6 * max(1.0, ts.Sg) and G.size == ts.n, (s, m)
            for M in MODEL_NAMES:
                if M == "EVA":
                    continue
                pc = np.nan_to_num(TSP[(M, m)].astype(np.float64))
                Pm = np.where(pos >= 0, pc[np.maximum(pos, 0)], 0.0)
                rows.append(patch_row(M, m, s, G, Pm, Pm))
    pd.DataFrame(rows).to_csv(f"{OUT}/patch_sums_test.csv", index=False)
    assert PATCH_CHECK and max(PATCH_CHECK) < 1e-5, PATCH_CHECK
    print(f"patch sums: {len(rows)} rows, {len(TEST_IMGS_J)} test images; max |r_direct - r_sums| = {max(PATCH_CHECK):.2e}")
tick("patch sums")
```

```python
tp, tcode = np.unique(DJ["patient"][J_TE], return_inverse=True)
Wt = np.random.default_rng(SEED + 12).multinomial(len(tp), np.full(len(tp), 1 / len(tp)), size=N_BOOT).astype(np.float64)
R4F = stats_fn(IMG_TEJ, tcode, len(tp), Wt, True)
img_pat = pd.Series(pd.Series(tcode, index=IMG_TEJ).groupby(level=0).first().reindex(TEST_IMGS_J).values, index=name_te)
cols6 = ["n", "Sg", "Sp", "Sgg", "Spp", "Sgp"]
EXCL = "estrogen_receptor_alpha"
BOOT, ROWS = {}, []

def record(M, metric, m, pt, bv, **extra):
    BOOT[(M, metric, m)] = (pt, bv)
    lo, hi = ci(bv)
    ROWS.append(dict(model=M, metric=metric, marker=m, value=pt, lo=lo, hi=hi, **extra))

for (M, m), df in SUMS_TE.groupby(["model", "marker"], sort=False):
    pc = img_pat.reindex(df.image).values.astype(np.int64)
    per_pat = np.stack([np.bincount(pc, df[c].values, len(tp)) for c in cols6])
    record(M, "pooled pixel r", m, float(r_from(*per_pat.sum(1))), r_from(*(Wt @ per_pat.T).T))
    ri = r_from(*(df[c].values for c in cols6))
    ok = np.isfinite(ri)
    sr = np.bincount(pc[ok], ri[ok], len(tp))
    cn = np.bincount(pc[ok], minlength=len(tp)).astype(np.float64)
    record(M, "per-image pixel r", m, float(sr.sum() / cn.sum()), (Wt @ sr) / (Wt @ cn), n_images=int(ok.sum()))
    pt, bv = R4F(TRUTHJ[m][J_TE], TSP[(M, m)])
    record(M, "R4", m, pt, bv)

PM = pd.DataFrame(ROWS)
PM.to_csv(f"{OUT}/metrics_per_marker.csv", index=False)
MEAN = PM[PM.marker != EXCL].groupby(["model", "metric"]).value.mean().unstack()
print("mean over 10 markers (ERα excluded), Jackson test patients:")
print(MEAN.round(3).to_string())
tick("metrics done")
json.dump({"failed_blocks": FAILED, "minutes": (time.time() - T0) / 60, "n_test_cells": int(len(J_TE)),
           "n_test_images": int(len(TEST_IMGS_J)), "n_test_patients": int(len(tp))}, open(f"{OUT}/run_info.json", "w"))
print("failed blocks:", FAILED if FAILED else "none")
```
